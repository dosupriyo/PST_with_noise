```python
def computational_basis(n, k):
    v = np.matrix([0]*n).transpose()
    v[k] = 1
    return(v)
```

```python
def ket_vector(n, v):
    ket = [0] * n
    ket[v] = 1
    return np.matrix(ket).transpose()
```

```python
def initial_state(G, s):
    d = G.degree(s)
    H = G.to_directed()
    E = list(H.edges())
    m = len(E)
    z = np.matrix(np.zeros((m, 1)))
    out_edges_of_s = H.out_edges(s)
    for e in out_edges_of_s:
        i = E.index(e)
        z = z + computational_basis(m, i)
    return(z/math.sqrt(d))
```

```python
def target_state(G, r):
    d = G.degree(r)
    H = G.to_directed()
    E = list(H.edges())
    m = len(E)
    z = np.matrix(np.zeros((m, 1)))
    out_edges_of_r = H.out_edges(r)
    for e in out_edges_of_r:
        i = E.index(e)
        z = z + computational_basis(m, i)
    return(z/math.sqrt(d))
```

```python
phi = target_state(G, r)
phi
```

```python
psi = initial_state(G, s)
psi
```

```python
def Grover(G, v):
    n = G.order()
    d = G.degree(v)
    psi = np.matrix(np.zeros((d, 1)))
    for i in range(d):
        psi = psi + computational_basis(d, i)
    psi = psi/math.sqrt(d)
    I_d = np.matrix(np.eye(d))
    G = 2 * psi * psi.transpose() - I_d
    return(np.matrix(G))
```

```python
def coin(G, s, r):
    L = []
    for v in G.nodes():
        if v == s or v == r:
            L.append(-1 * Grover(G, v))
        else:
            L.append(Grover(G, v))
    return(np.matrix(lin.block_diag(*L)))
```

```python
def shift_operator(G):
    m = G.number_of_edges()
    S = np.zeros((2*m , 2*m))
    V = list(G.nodes())
    Directed_edges = []
    for v in V:
        Temp = np.sort(list(G.neighbors(v)))
        for u in Temp:
            Directed_edges.append((v,u))
    for e in Directed_edges:
        i = Directed_edges.index(e)
        f = (e[1], e[0])
        j = Directed_edges.index(f)
        S[i][j] = 1
    return(np.matrix(S))
```

```python
def evol_operator(G, s, r):
    S = shift_operator(G)
    C = coin(G, s , r)
    T = S * C
    return(np.matrix(T))
```

```python
U = evol_operator(G, s, r)
sym.Matrix(U)
```

```python
T = 100
```

```python
def evolve_vector_t_step(U, psi, t):
    return (U**t)*psi
```

##### Fidelity for state vectors

Given state vectors $|a>$ and $|b>$ the fidelity is given by
$$F(a, b) = |<a | b >|^2.$$

```python
def fidelity_with_vectors(a, b):
    f = abs(a.transpose() * b)**2
    return(f.item())
```

```python
a = phi
F_without_noise = []
for t in np.arange(0, T+1):
    b = evolve_vector_t_step(U, psi, t)
    f = fidelity_with_vectors(a, b)
    F_without_noise.append(f)
```

```python
plt.figure(figsize = (20, 5))
plt.plot(T_axis, F_without_noise)
plt.show()
```

## State transfer with noise

##### Noise function
$$
\Lambda(t) = e^{-\gamma t} \left[ \cos\left(\left[ \sqrt{\left( \frac{2a}{\gamma} \right)^2 - 1}\right] \gamma t \right) + \frac{\sin\left(\left[ \sqrt{\left( \frac{2a}{\gamma} \right)^2 - 1}\right] \gamma t \right)}{\sqrt{\left( \frac{2a}{\gamma} \right)^2 - 1}} \right]
$$


```python
def Lambda_function(a, gamma, t):
    l1 = cmath.cos(np.sqrt((2*a/gamma)**2-1)*gamma*t)
    l2 = l1 + (cmath.sin(np.sqrt((2*a/gamma)**2-1)*gamma*t))/(np.sqrt((2*a/gamma)**2-1))
    lambda_t = cmath.exp(-gamma*t)*l2
    return(lambda_t)
```

```python
a = .1
gamma = 0.01
```

```python
Lambda_function_value = []
for t in range(T):
    Lambda_function_value.append(Lambda_function(a, gamma, t))
plt.plot(range(T), Lambda_function_value)
plt.xlabel('Time')
plt.ylabel('Lambda function value')
plt.title('Lambda Function Over Time')
plt.show()
```

##### Weyl operator
 $$U_{u, u} = I_d,$$
$$U_{u, u_0} = \sum_{k = 0}^{d - 1} \exp\left(\frac{2\pi \iota u k}{d}\right) |k> <(k + u_0) mod d|$$


```python
def w_coeff(d, u, k):
    return(cmath.exp((2j * cmath.pi * u * k )/ (d)))
```

```python
def Weyl(G, u, u_0):
    m = G.number_of_edges()
    z = np.matrix(np.zeros((2 * m, 2 * m)))
    for k in range(2 * m):
        w = w_coeff(2 * m, u, k)
        z = z + w * ket_vector(2 * m, k) * ket_vector(2 * m, (k + u_0) % (2 * m)).transpose()
    return(z)
```

```python
u = 1
u_0 = 0
W = Weyl(G, u, u_0)
np.around(W*np.conjugate(W).transpose(), 4)
```



##### Kraus opeartors

```python
def K1(G, a, gamma, t):
    m = G.number_of_edges()
    I = np.matrix(np.eye((2 * m)))
    Lambda = Lambda_function(a, gamma, t)
    z = np.sqrt((1 + Lambda)/2)*I
    return(z)
```

```python
def K2(G, a, gamma, t):
    m = G.number_of_edges()
    Lambda = Lambda_function(a, gamma, t)
    z = np.sqrt((1-Lambda)/2)*Weyl(G, u, u_0)
    return(z)
```

```python
t = 10
K_2 = K2(G, a, gamma, t)
K_1 = K1(G, a, gamma, t)
np.around(K_1*np.conjugate(K_1).transpose() + K_2*np.conjugate(K_2).transpose(), 4)
```

### Evolution of quantum walk

$$ \rho(t) = \sum_{j} K_{j}(t) (U^t\rho(0) U^{t\dagger}) K_{j}^{\dagger}(t) $$

```python
def evolve_matrix_t_step(G, a, gamma, t, IS, s, r):
    K_1 = K1(G, a, gamma, t)  
    K_2 = K2(G, a, gamma, t)
    U = evol_operator(G, s, r)
    rho = (U**t) * IS * np.conjugate(U**t).transpose()
    rho_t = K_1 * rho * np.conjugate(K_1).transpose() + K_2 * rho * np.conjugate(K_2).transpose()
    return rho_t
```

```python
IS = psi * psi.transpose()
IS
```

##### Fidelity between two states represented by density matrices

$$ F(\rho, \sigma)  =  (Tr(\sqrt{\sqrt{\sigma}\rho\sqrt{\sigma}}))^2 $$

```python
def fidelity(rho, sigma):
    fidelity = np.trace(lin.sqrtm((lin.sqrtm(sigma) * rho * lin.sqrtm(sigma))))
    fidelity = np.abs(fidelity) ** 2
    return fidelity
```

```python
rho = phi * phi.transpose()
F_with_noise = []
for t in np.arange(0, T+1):
    sigma = np.around(evolve_matrix_t_step(G, a, gamma, t, IS, s, r), 10)
    f = fidelity(rho, sigma)
    F_with_noise.append(f)
```

```python
plt.figure(figsize=(20, 5))
plt.grid(True, color='gray', linestyle='--', linewidth=0.5)
plt.plot(T_axis, F_without_noise, label="without noise", color = "blue")
plt.plot(T_axis, F_with_noise, label="with noise", color = "red")
plt.xticks(list(np.arange(0, T+5, 5)), fontsize = 13, fontweight='bold')
plt.yticks(fontsize=13, fontweight='bold')
plt.legend(loc=(0.85, .8), fontsize=13)
plt.text(25, 1.1, r"Vertices of receiver and sender are $\mathbf{s = " + str(s) + "},\ \mathbf{r = " + str(r) + "}$, respectively.",
fontsize=13,fontweight='bold')
plt.xlabel("Time steps of quantum walk", fontsize=13,fontweight='bold')
plt.ylabel("Fidelity between initial and final states", fontsize=13, fontweight='bold')
plt.show()
```
